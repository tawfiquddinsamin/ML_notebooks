###Removal of stop words

```python
def remove_stopwords(text):
  x=str(text).split()
  new_text=[]
  for word in x:
    if word not in stop_words:
      new_text.append(word)
  clean_text=' '.join(new_text)
  return clean_text

cleaned_texts=df['content'].apply(remove_stopwords)
df['clean_text']=cleaned_texts
df['content'][0]
```

###Dealing with contractions

```python
contractions = {
"বি.দ্র ": "বিশেষ দ্রষ্টব্য",
"ড.": "ডক্টর",
"ডা.": "ডাক্তার",
"ইঞ্জি:": "ইঞ্জিনিয়ার",
"রেজি:": "রেজিস্ট্রেশন",
"মি.": "মিস্টার",
"মু.": "মুহাম্মদ",
"মো.": "মোহাম্মদ",
}

def replace_contractions(text):
    new_text = []
    x=str(text).split()
    for w in x:
      if w in contractions:
        new_text.append(contractions[w])
      else:
        new_text.append(w)
    clean_text=' '.join(new_text)
    return clean_text
```

###punctuation and english word removal

```python
import re
import string
def remove_punctuations(text):
  x=text
  punctuations=r"[(),$%^&*+={}:\"|\'\~`<>/,€¦!?½£¶™¼©⅐⅑⅒⅓⅔⅕⅖⅗⅘⅙⅚⅛⅜⅝⅞⅟↉¤¿º;-]+"
  punctuation_sequence= punctSeq   = r"['\"“”‘’]+|[.?!,…]+|[:;]+"
  whitespace = re.compile(r"[\s\u0020\u00a0\u1680\u180e\u202f\u205f\u3000\u2000-\u200a]+", re.UNICODE)
  full_stop=r"\u0964"
  english_chars = r"[a-zA-Z0-9]"
  x=re.sub(punctuations," ",x)
  x=re.sub(punctuation_sequence," ",x)
  x=re.sub(whitespace," ",x)
  x=re.sub(full_stop," ",x)
  x=re.sub(english_chars," ",x)
  return x
```

```python
x=df['content'].apply(replace_contractions)
df['content']=x
df['content']=df['content'].apply(remove_punctuations)

df.head()
```

Random Forest

```python
from sklearn.ensemble import RandomForestClassifier
from sklearn.model_selection import KFold
from sklearn.metrics import accuracy_score, precision_score, recall_score, f1_score
import numpy as np
import time
import tracemalloc

# KFold setup
rf_kf = KFold(n_splits=5, shuffle=True, random_state=42)

# Metric storage
rf_acc_list = []
rf_prec_list = []
rf_rec_list = []
rf_f1_list = []
rf_time_list = []
rf_memory_list = []

rf_model = RandomForestClassifier(random_state=42)
for rf_train_idx, rf_test_idx in rf_kf.split(x_train):

    # Correcting indexing for sparse matrices (x_train, x_test)
    rf_X_tr = x_train[rf_train_idx] # Removed .iloc
    rf_X_te = x_train[rf_test_idx]  # Removed .iloc
    rf_y_tr = y_train.iloc[rf_train_idx]
    rf_y_te = y_train.iloc[rf_test_idx]



    # Memory tracking start
    tracemalloc.start()

    # Time tracking
    rf_start_time = time.time()
    rf_model.fit(rf_X_tr, rf_y_tr)
    rf_end_time = time.time()

    rf_current_mem, rf_peak_mem = tracemalloc.get_traced_memory()
    tracemalloc.stop()

    # Prediction
    rf_y_pred = rf_model.predict(rf_X_te)

    # Metrics
    rf_acc_list.append(accuracy_score(rf_y_te, rf_y_pred))
    rf_prec_list.append(precision_score(rf_y_te, rf_y_pred))
    rf_rec_list.append(recall_score(rf_y_te, rf_y_pred))
    rf_f1_list.append(f1_score(rf_y_te, rf_y_pred))

    # Time & Memory
    rf_time_list.append(rf_end_time - rf_start_time)
    rf_memory_list.append(rf_peak_mem / 10**6)  # MB

# Final averages
rf_acc_mean = np.mean(rf_acc_list)
rf_prec_mean = np.mean(rf_prec_list)
rf_rec_mean = np.mean(rf_rec_list)
rf_f1_mean = np.mean(rf_f1_list)
rf_time_mean = np.mean(rf_time_list)
rf_memory_mean = np.mean(rf_memory_list)

print("===== Random Forest (5-Fold CV) =====")
print(f"RF Accuracy: {rf_acc_mean:.4f}")
print(f"RF Precision: {rf_prec_mean:.4f}")
print(f"RF Recall: {rf_rec_mean:.4f}")
print(f"RF F1-score: {rf_f1_mean:.4f}")
print(f"RF Avg Training Time: {rf_time_mean:.4f} sec")
print(f"RF Avg Memory Usage: {rf_memory_mean:.4f} MB")
```

```python
from sklearn.metrics import classification_report
rf_clreport=classification_report(y_test,rf_model.predict(x_test))
print(rf_clreport)
```

```python
from sklearn.neighbors import KNeighborsClassifier
from sklearn.model_selection import KFold
from sklearn.metrics import accuracy_score, precision_score, recall_score, f1_score
import numpy as np
import time
import tracemalloc

knn_kf = KFold(n_splits=5, shuffle=True, random_state=42)

knn_acc_list, knn_prec_list, knn_rec_list, knn_f1_list = [], [], [], []
knn_time_list, knn_memory_list = [], []


knn_model = KNeighborsClassifier(n_neighbors=2)
for knn_train_idx, knn_test_idx in knn_kf.split(x_train):

    knn_X_tr = x_train[knn_train_idx]
    knn_X_te = x_train[knn_test_idx]
    knn_y_tr = y_train.iloc[knn_train_idx]
    knn_y_te = y_train.iloc[knn_test_idx]


    tracemalloc.start()
    knn_start = time.time()
    knn_model.fit(knn_X_tr, knn_y_tr)
    knn_end = time.time()
    _, knn_peak = tracemalloc.get_traced_memory()
    tracemalloc.stop()

    knn_pred = knn_model.predict(knn_X_te)

    knn_acc_list.append(accuracy_score(knn_y_te, knn_pred))
    knn_prec_list.append(precision_score(knn_y_te, knn_pred))
    knn_rec_list.append(recall_score(knn_y_te, knn_pred))
    knn_f1_list.append(f1_score(knn_y_te, knn_pred))

    knn_time_list.append(knn_end - knn_start)
    knn_memory_list.append(knn_peak / 10**6)

print("===== KNN =====")
print(f"KNN Accuracy: {np.mean(knn_acc_list):.4f}")
print(f"KNN F1: {np.mean(knn_f1_list):.4f}")
print(f"KNN Time: {np.mean(knn_time_list):.4f}")
print(f"KNN Memory: {np.mean(knn_memory_list):.4f}")
```

### Run time bar graph

```python
import matplotlib.pyplot as plt
from matplotlib import rcParams
from matplotlib.colors import LinearSegmentedColormap
import numpy as np


times = [
    rf_time_mean,
    dt_time_mean,
    lr_time_mean,
    np.mean(knn_time_list),
    np.mean(svm_time_list),
    np.mean(nb_time_list),
]
algo_list = [
    "Random Forest",
    "Decision Tree",
    "Logistic Regression",
    "K-Nearest Neighbors",
    "Support Vector Machine",
    "Naïve Bayes",
]

order     = np.argsort(times)[::-1]
times     = [times[i]     for i in order]
algo_list = [algo_list[i] for i in order]


BG   = "#F7F5F0"
PANEL= "#EDEAE2"
INK  = "#1C1A17"
MID  = "#6B6560"
RULE = "#C8C3B8"

rcParams.update({
    "font.family":      "serif",
    "font.serif":       ["Georgia", "DejaVu Serif", "Times New Roman"],
    "figure.facecolor": BG,
    "axes.facecolor":   PANEL,
    "text.color":       INK,
    "axes.labelcolor":  INK,
    "xtick.color":      MID,
    "ytick.color":      INK,
    "axes.edgecolor":   RULE,
    "grid.color":       RULE,
    "grid.linewidth":   0.6,
})

cmap   = LinearSegmentedColormap.from_list(
    "time", ["#4A90B8", "#6CB89A", "#E9C46A", "#E07B54", "#C0392B"], N=256
)
norm   = plt.Normalize(vmin=min(times), vmax=max(times))
colors = [cmap(norm(v)) for v in times]

fig, ax = plt.subplots(figsize=(11, 6.5), facecolor=BG)
ax.set_facecolor(PANEL)

y_pos = np.arange(len(algo_list))
bars  = ax.barh(y_pos, times, color=colors, height=0.62,
                edgecolor="white", linewidth=0.8, zorder=3)

max_val = max(times)
for bar, val in zip(bars, times):
    w      = bar.get_width()
    inside = w > max_val * 0.18
    xpos   = w - max_val * 0.012 if inside else w + max_val * 0.012
    ha     = "right" if inside else "left"
    color  = "white"  if inside else INK


    label  = f"{val*1000:.1f} ms" if max_val < 1 else f"{val:.3f} s"
    ax.text(xpos, bar.get_y() + bar.get_height() / 2,
            label,
            va="center", ha=ha,
            fontsize=10, fontfamily="serif",
            color=color, fontweight="bold")



ref_step = round(max_val / 4, 3)
if ref_step == 0:
    ref_step = max_val / 4
refs = [round(ref_step * i, 4) for i in range(1, 5) if ref_step * i < max_val]
for xref in refs:
    ax.axvline(xref, color=RULE, linewidth=0.8, linestyle="--",
               zorder=2, alpha=0.9)
    label = f"{xref*1000:.0f}ms" if max_val < 1 else f"{xref:.2f}s"
    ax.text(xref, len(algo_list) - 0.02, label,
            va="bottom", ha="center",
            fontsize=7.5, color=MID, fontfamily="serif")

ax.set_yticks(y_pos)
ax.set_yticklabels(algo_list, fontsize=11.5, fontfamily="serif")
unit = "ms" if max_val < 1 else "seconds"
ax.set_xlabel(f"Mean Training Time  ({unit})", fontsize=11,
              fontfamily="serif", style="italic", labelpad=10)
ax.set_xlim(0, max_val * 1.18)
ax.set_ylim(-0.6, len(algo_list) - 0.4)
ax.invert_yaxis()
ax.spines[["top", "right"]].set_visible(False)
ax.spines["left"].set_color(RULE)
ax.spines["bottom"].set_color(RULE)
ax.tick_params(axis="x", labelsize=9, labelcolor=MID, length=3)
ax.tick_params(axis="y", length=0)
ax.set_xticks([])


for rank, bar in enumerate(bars, 1):
    ax.text(-max_val * 0.015, bar.get_y() + bar.get_height() / 2,
            f"#{rank}",
            va="center", ha="right",
            fontsize=8, fontfamily="serif",
            color=MID, style="italic")

sm = plt.cm.ScalarMappable(cmap=cmap, norm=norm)
sm.set_array([])
cbar = fig.colorbar(sm, ax=ax, orientation="vertical",
                    shrink=0.55, aspect=20, pad=0.02)
cbar.set_label(f"Time  ({unit})", fontsize=9, fontfamily="serif",
               color=MID, style="italic")
cbar.ax.tick_params(labelsize=8, color=MID)
plt.setp(cbar.ax.yaxis.get_ticklabels(), color=MID, fontfamily="serif")
cbar.outline.set_edgecolor(RULE)


fig.text(0.5, 0.97,
         "Training Time Comparison of Machine Learning Classifiers",
         ha="center", va="top",
         fontsize=15, fontfamily="serif", fontweight="bold", color=INK)
fig.text(0.5, 0.925,
         "Mean training time per cross-validation fold · sorted slowest → fastest",
         ha="center", va="top",
         fontsize=9.5, fontfamily="serif", style="italic", color=MID)
fig.add_artist(plt.Line2D([0.08, 0.92], [0.912, 0.912],
               transform=fig.transFigure, color=RULE, linewidth=0.9))


plt.tight_layout(rect=[0, 0.04, 1, 0.90])
plt.show()
```
